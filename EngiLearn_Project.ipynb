```python

import requests
from pathlib import Path

# Load the cleaned text extracted from your engineering PDF
#text_path = Path("C:\Users\For YOU\Downloads\cleaned_text.txt")

text_path = Path(r"C:\Users\For YOU\Downloads\cleaned_text.txt")


if not text_path.exists():
    raise FileNotFoundError(
        "cleaned_text.txt was not found. "
        "Check that the notebook is in your EngiLearn project folder."
    )

document_text = text_path.read_text(encoding="utf-8")
print("Characters loaded:", len(document_text))

# Start with a limited section for the first test
sample_text = document_text[:6000]

prompt = f"""
You are an engineering learning assistant.

Based ONLY on the document content below:
1. Write a summary in simple English.
2. List five important concepts.
3. Explain difficult technical terms in beginner-friendly language.

Do not invent facts that are not supported by the document.

DOCUMENT CONTENT:
{sample_text}
"""

OLLAMA_URL = "http://192.168.137.192:11434"

response = requests.post(
    f"{OLLAMA_URL}/api/generate",
    json={
        "model": "qwen2.5:1.5b",
        "prompt": prompt,
        "stream": False
    },
    timeout=300
)

response.raise_for_status()
summary = response.json()["response"]

print(summary)
```
